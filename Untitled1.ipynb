```python
import numpy as np
import pandas as pd

# Generate a time series with random values and a monthly frequency
ts = pd.Series(np.random.randn(4), index=pd.date_range('1/1/2024', periods=4, freq='ME'))
print("Original Time Series:")
print(ts)

# Lagging by 2 periods (shift data backward)
print("\nLagged Time Series (shifted by 2 periods):")
print(ts.shift(2))

# Leading by 2 periods (shift data forward)
print("\nLed Time Series (shifted by -2 periods):")
print(ts.shift(-2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming your data is in a CSV file
# Replace the path with the actual path to your file
# data = pd.read_csv("yahoo_stock.csv")

# Example DataFrame setup (you should replace this with your actual data)
data = pd.DataFrame({
    'Date': ['2015-11-23', '2015-11-24', '2015-11-25', '2015-11-26', '2015-11-27'],
    'High': [2095.61, 2094.12, 2093.00, 2093.00, 2093.29]
})

# Convert 'Date' to datetime
data['Date'] = pd.to_datetime(data['Date'])

# Set 'Date' column as the index
data.set_index('Date', inplace=True)

# Resample the data to get monthly averages of 'High' column using 'ME' (Month End)
avr_month = data[['High']].resample('ME').mean()

# Plot the results using Matplotlib
plt.figure(figsize=(15, 8))
plt.plot(avr_month.index, avr_month['High'], label='Monthly Average High')
plt.xlabel('Date')
plt.ylabel('Average High Value')
plt.title('Monthly Average of High Values')
plt.legend()
plt.show()
```

```python
import yfinance as yf
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Step 1: Download stock data (e.g., Apple stock)
ticker = 'AAPL'  # Apple stock symbol
data = yf.download(ticker, start='2020-01-01', end='2023-12-31')

# Step 2: Resample data to monthly average of 'High' values
avr_month = data[['High']].resample('M').mean()

# Step 3: Plotting the average monthly 'High' values
plt.figure(figsize=(15, 8))
sns.lineplot(x=avr_month.index, y=avr_month['High'])
plt.title(f'Average Monthly High Values for {ticker}')
plt.xlabel('Date')
plt.ylabel('High Value')
plt.show()
```
