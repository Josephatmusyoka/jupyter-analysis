```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Data
data = {
    'Age': [65, 50, 80, 25, 36, 43, 64, 73, 19, 22, 36, 29, 34, 62, 68, 91, 42, 14, 57, 61],
    'Weight': [90, 78, 94, 70, 80, 104, 83, 87, 69, 98, 114, 92, 64, 72, 50, 70, 110, 84, 68, 94],
    'BP (Systolic)': [130, 125, 140, 122, 130, 148, 128, 135, 121, 149, 158, 130, 126, 132, 129, 110, 143, 162, 118, 131],
    'BP (Diastolic)': [90, 86, 98, 84, 88, 102, 82, 87, 81, 96, 98, 88, 83, 89, 78, 74, 86, 99, 98, 87],
    'Height': [1.95, 1.7, 1.86, 1.94, 1.78, 1.88, 1.9, 1.79, 1.78, 1.7, 1.84, 2.1, 1.9, 1.78, 1.74, 1.68, 1.82, 1.88, 1.72, 1.85]
}

df = pd.DataFrame(data)

# Standardizing the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# PCA
pca = PCA()
pca.fit(scaled_data)

# Explained variance
explained_variance = pca.explained_variance_ratio_

# Components (Eigenvectors)
components = pca.components_

print("Explained Variance Ratio:")
print(explained_variance)

print("\nPrincipal Components:")
print(components)

# Plot explained variance ratio
plt.figure(figsize=(8,6))
plt.bar(range(1, len(explained_variance) + 1), explained_variance, alpha=0.7, align='center', color='b')
plt.xlabel('Principal Components')
plt.ylabel('Explained Variance Ratio')
plt.title('Explained Variance Ratio by Principal Components')
plt.show()

# Project the data onto the first two principal components
pca_2d = PCA(n_components=2)
data_2d = pca_2d.fit_transform(scaled_data)

# DataFrame for 2D projection
pca_df = pd.DataFrame(data_2d, columns=['PC1', 'PC2'])

# Display first few rows of 2D projected data
print(pca_df.head())

# 2D scatter plot of the projected data
plt.figure(figsize=(8,6))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c='r', alpha=0.7)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('2D Projection of Data using PCA')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Step 1: Create the DataFrame
data = {
    'Age': [65, 50, 80, 25, 36, 43, 64, 73, 19, 22, 36, 29, 34, 62, 68, 91, 42, 14, 57, 61],
    'Weight': [90, 78, 94, 70, 80, 104, 83, 87, 69, 98, 114, 92, 64, 72, 50, 70, 110, 84, 68, 94],
    'BP (Systolic)': [130, 125, 140, 122, 130, 148, 128, 135, 121, 149, 158, 130, 126, 132, 129, 110, 143, 162, 118, 131],
    'BP (Diastolic)': [90, 86, 98, 84, 88, 102, 82, 87, 81, 96, 98, 88, 83, 89, 78, 74, 86, 99, 98, 87],
    'Height': [1.95, 1.7, 1.86, 1.94, 1.78, 1.88, 1.9, 1.79, 1.78, 1.7, 1.84, 2.1, 1.9, 1.78, 1.74, 1.68, 1.82, 1.88, 1.72, 1.85]
}

df = pd.DataFrame(data)

# Step 2: Standardize the data (important for PCA)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Step 3: Perform PCA
pca = PCA()
pca.fit(scaled_data)

# Step 4: Get the explained variance ratio (the percentage of variance explained by each component)
explained_variance = pca.explained_variance_ratio_

# Step 5: Get the principal components (eigenvectors)
components = pca.components_

# Display the results
print("Explained Variance Ratio for each Principal Component:")
print(explained_variance)

print("\nPrincipal Components (Eigenvectors):")
print(components)

# Step 6: Plot the Explained Variance Ratio
plt.figure(figsize=(8, 6))
plt.bar(range(1, len(explained_variance) + 1), explained_variance, alpha=0.7, align='center', color='b')
plt.xlabel('Principal Components')
plt.ylabel('Explained Variance Ratio')
plt.title('Explained Variance Ratio by Principal Components')
plt.show()

# Step 7: Reduce the dimensionality to 2 components and project the data
pca_2d = PCA(n_components=2)
data_2d = pca_2d.fit_transform(scaled_data)

# Create a DataFrame for the 2D projected data
pca_df = pd.DataFrame(data_2d, columns=['PC1', 'PC2'])

# Display the first few rows of the projected data
print("\nProjected Data (First 2 Principal Components):")
print(pca_df.head())

# Step 8: Plot the 2D projection of the data
plt.figure(figsize=(8, 6))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c='r', alpha=0.7)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('2D Projection of Data using PCA')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load the data
data = {'Age': [65, 50, 70, 10, 83, 69, 26, 84, 10, 14], 
        'Weight': [90, 78, 98, 102, 87, 83, 83, 99, 96, 99], 
        'BP(systolic)': [130, 125, 140, 143, 128, 135, 121, 129, 110, 162], 
        'BP(Diastolic)': [90, 86, 98, 86, 82, 88, 81, 78, 74, 99], 
        'Height': [1.95, 1.78, 1.86, 1.94, 1.79, 1.96, 1.78, 1.75, 1.68, 1.89]}

df = pd.DataFrame(data)

# Standardize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Perform PCA
pca = PCA()
pca.fit(scaled_data)

# Get principal components
principal_components = pca.components_

# Print the principal components
print("Principal Components:")
print(principal_components)

# Explained variance ratio
explained_variance_ratio = pca.explained_variance_ratio_
print("\nExplained Variance Ratio:")
print(explained_variance_ratio)

# Cumulative explained variance
cumulative_variance_ratio = np.cumsum(explained_variance_ratio)
print("\nCumulative Explained Variance Ratio:")
print(cumulative_variance_ratio)
```
