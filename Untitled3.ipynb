```python
import pandas as pd
import matplotlib.pyplot as plt
from fbprophet import Prophet

# Load the dataset
df = pd.read_csv(r'C:\Users\STANZO\Downloads\orders_export-1736754434171.csv')

# Convert date columns to datetime
df['Created At'] = pd.to_datetime(df['Created At'])

# Total Sales and Average Order Value
total_sales = df['Paid Price'].sum()
average_order_value = df['Paid Price'].mean()

print("Total Sales:", total_sales)
print("Average Order Value:", average_order_value)

# Sales by Product
top_products = df.groupby('Item Name')['Paid Price'].sum().sort_values(ascending=False).head(5)
print("Top 5 Products by Sales:\n", top_products)

# Customer Insights
unique_customers = df['Customer First Name'].nunique()
repeat_customers = df.groupby(['Customer First Name', 'Customer Last Name']).size().reset_index(name='order_count')
repeat_customers = repeat_customers[repeat_customers['order_count'] > 1].shape[0]
print("Number of Unique Customers:", unique_customers)
print("Number of Repeat Customers:", repeat_customers)

# Order Status Analysis
order_status_distribution = df['Status'].value_counts(normalize=True) * 100
print("Order Status Distribution (%)\n", order_status_distribution)

# Shipping and Delivery Performance
average_shipping_cost = df['Shipping Fee'].mean()
print("Average Shipping Cost:", average_shipping_cost)

# Trends Over Time
sales_trends = df.set_index('Created At').resample('M')['Paid Price'].sum()
print("Sales Trends Over Time:\n", sales_trends)

# Plot sales trends (requires matplotlib)
sales_trends.plot(kind='line', title='Monthly Sales Trends')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()

# --- Advanced Analysis ---

# 1. Customer Segmentation (Spending Categories)
customer_spending = df.groupby(['Customer First Name', 'Customer Last Name'])['Paid Price'].sum().reset_index()

def classify_spending(spending):
    if spending > customer_spending['Paid Price'].quantile(0.75):
        return 'High'
    elif spending > customer_spending['Paid Price'].quantile(0.25):
        return 'Medium'
    else:
        return 'Low'

customer_spending['Spending Category'] = customer_spending['Paid Price'].apply(classify_spending)
print("\nCustomer Segmentation:\n", customer_spending['Spending Category'].value_counts())

# 2. Customer Lifetime Value (CLV)
customer_freq = df.groupby(['Customer First Name', 'Customer Last Name']).size().reset_index(name='Purchase Frequency')
customer_clv = pd.merge(customer_spending, customer_freq, on=['Customer First Name', 'Customer Last Name'])
customer_clv['AOV'] = customer_clv['Paid Price'] / customer_clv['Purchase Frequency']
customer_clv['First Order Date'] = df.groupby(['Customer First Name', 'Customer Last Name'])['Created At'].min().reset_index()['Created At']
customer_clv['Lifespan (Months)'] = (pd.to_datetime('today') - customer_clv['First Order Date']).dt.days / 30
customer_clv['CLV'] = customer_clv['AOV'] * customer_clv['Purchase Frequency'] * customer_clv['Lifespan (Months)']
print("\nCustomer Lifetime Value (CLV):\n", customer_clv[['Customer First Name', 'Customer Last Name', 'CLV']].head())

# 3. RFM Analysis
recency = df.groupby(['Customer First Name', 'Customer Last Name'])['Created At'].max().reset_index()
recency['Recency'] = (pd.to_datetime('today') - recency['Created At']).dt.days

frequency = df.groupby(['Customer First Name', 'Customer Last Name']).size().reset_index(name='Frequency')

monetary = df.groupby(['Customer First Name', 'Customer Last Name'])['Paid Price'].sum().reset_index()

rfm = pd.merge(recency, frequency, on=['Customer First Name', 'Customer Last Name'])
rfm = pd.merge(rfm, monetary, on=['Customer First Name', 'Customer Last Name'])

rfm['Recency_Score'] = pd.qcut(rfm['Recency'], 5, labels=False)
rfm['Frequency_Score'] = pd.qcut(rfm['Frequency'], 5, labels=False)
rfm['Monetary_Score'] = pd.qcut(rfm['Paid Price'], 5, labels=False)

rfm['RFM_Score'] = rfm['Recency_Score'] + rfm['Frequency_Score'] + rfm['Monetary_Score']
print("\nRFM Segmentation:\n", rfm.head())

# 4. Order Volume by Time of Day
df['Hour'] = df['Created At'].dt.hour
order_volume_by_hour = df.groupby('Hour').size()
order_volume_by_hour.plot(kind='bar', title='Order Volume by Hour of the Day', figsize=(10,6))
plt.xlabel('Hour of the Day')
plt.ylabel('Order Volume')
plt.xticks(rotation=0)
plt.show()

# 5. Sales Forecasting (Time Series with Prophet)
sales_trends_df = sales_trends.reset_index().rename(columns={'Created At': 'ds', 'Paid Price': 'y'})

model = Prophet()
model.fit(sales_trends_df)

future = model.make_future_dataframe(sales_trends_df, periods=12, freq='M')
forecast = model.predict(future)

model.plot(forecast)
plt.title('Sales Forecast for the Next 12 Months')
plt.show()
```
